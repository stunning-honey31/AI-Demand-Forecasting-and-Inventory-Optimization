```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

df = pd.read_csv(r"C:\Users\G VARSHIT HARIPRASAD\Desktop\Infotact Solutions Project\Month 3 - May\AI-Demand-Forecasting-and-Inventory-Optimization\data\raw\train.csv")

df.head()
```

```python
df.describe()
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Weekly Demand Trend

weekly_sales = df.groupby('week')['num_orders'].sum()

plt.figure(figsize=(14,6))

plt.plot(weekly_sales)

plt.title("Weekly Food Demand Trend")
plt.xlabel("Week")
plt.ylabel("Total Orders")

plt.grid()

plt.show()
```

```python

top_meals = df.groupby('meal_id')['num_orders'].sum() \
              .sort_values(ascending=False) \
              .head(10)

plt.figure(figsize=(12,6))

top_meals.plot(kind='bar')

plt.title("Top 10 Selling Meals")
plt.xlabel("Meal ID")
plt.ylabel("Total Orders")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['num_orders'], bins=50, kde=True)

plt.title("Distribution of Food Orders")
plt.xlabel("Number of Orders")

plt.show()
```
