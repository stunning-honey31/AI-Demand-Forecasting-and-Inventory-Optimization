```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("../data/raw/train.csv")

# Display First Rows
df.head()
```

```python
weekly_orders = df.groupby('week')['num_orders'].sum()

plt.figure(figsize=(14,6))

plt.plot(
    weekly_orders.index,
    weekly_orders.values,
    linewidth=2
)

plt.title("Weekly Food Demand Trend")
plt.xlabel("Week")
plt.ylabel("Total Orders")

plt.grid()

plt.savefig("../outputs/plots/weekly_demand_trend.png")

plt.show()
```

```python
top_meals = df.groupby('meal_id')['num_orders'] \
              .sum() \
              .sort_values(ascending=False) \
              .head(10)

plt.figure(figsize=(12,6))

top_meals.plot(kind='bar')

plt.title("Top 10 Most Ordered Meals")
plt.xlabel("Meal ID")
plt.ylabel("Total Orders")

plt.savefig("../outputs/plots/top_10_meals.png")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.histplot(
    df['num_orders'],
    bins=50,
    kde=True
)

plt.title("Distribution of Food Orders")
plt.xlabel("Number of Orders")

plt.savefig("../outputs/plots/order_distribution.png")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=df['checkout_price'],
    y=df['num_orders']
)

plt.title("Checkout Price vs Number of Orders")
plt.xlabel("Checkout Price")
plt.ylabel("Number of Orders")

plt.savefig("../outputs/plots/checkout_price_vs_orders.png")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=df['base_price'],
    y=df['num_orders']
)

plt.title("Base Price vs Number of Orders")
plt.xlabel("Base Price")
plt.ylabel("Number of Orders")

plt.savefig("../outputs/plots/baseprice_price_vs_no__orders.png")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    df[['checkout_price', 'base_price', 'emailer_for_promotion',
        'homepage_featured', 'num_orders']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Feature Correlation Heatmap")

plt.savefig("../outputs/plots/correlation_heatmap.png")

plt.show()
```
